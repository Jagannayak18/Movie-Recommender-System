## Movie Recommender System 

#### Libraries

```python
import numpy as np
import pandas as pd
```

#### Dataset

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head()
movies.shape
movies.info()
```

```python
credits.head()
credits.shape
credits.info()
```

##### Merging the two datasets

```python
movies_list = movies.merge(credits, on='title')
movies_list.info()
```

#### Retrieving important columns

```python
movies_list_filtered = movies_list[['genres','keywords','overview','title','movie_id','cast','crew']]
movies_list_filtered.head()
```

#### Finding nulls

```python
movies_list_filtered.isnull().sum()
```

```python
movies_list_filtered.dropna(inplace=True)
movies_list_filtered.isnull().sum()
```

```python
import ast
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies_list_filtered['genres'] = movies_list_filtered['genres'].apply(convert)
movies_list_filtered['keywords'] = movies_list_filtered['keywords'].apply(convert)
movies_list_filtered['cast'] = movies_list_filtered['cast'].apply(convert)
movies_list_filtered.head()
```

```python
# To fetch only top 3 cast members
movies_list_filtered['cast'] = movies_list_filtered['cast'].apply(lambda x:x[0:3])
```

```python
# To fetch director from crew
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
```

```python
movies_list_filtered['crew'] = movies_list_filtered['crew'].apply(fetch_director)
movies_list_filtered.head()
```

```python
# To collapse spaces in between words
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies_list_filtered['genres'] = movies_list_filtered['genres'].apply(collapse)
movies_list_filtered['keywords'] = movies_list_filtered['keywords'].apply(collapse)
movies_list_filtered['cast'] = movies_list_filtered['cast'].apply(collapse)
movies_list_filtered['crew'] = movies_list_filtered['crew'].apply(collapse)
movies_list_filtered.head()
```

```python
#   To split overview into list of words
movies_list_filtered['overview'] = movies_list_filtered['overview'].apply(lambda x: x.split())
```

```python
#   To combine all into single tags column
movies_list_filtered['tags'] = movies_list_filtered['overview'] + movies_list_filtered['genres'] + movies_list_filtered['keywords'] + movies_list_filtered['cast'] + movies_list_filtered['crew']
movies_list_filtered.drop(columns=['overview','genres','keywords','cast','crew'], inplace=True)
movies_list_filtered.head()
```

```python
#   To convert list of words into single string
movies_list_filtered['tags'] = movies_list_filtered['tags'].apply(lambda x: " ".join(x))
movies_list_filtered.head()
```

```python
# Vectorization
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
# Creating  the  vectors
vectors = cv.fit_transform(movies_list_filtered['tags']).toarray()
vectors.shape
```

```python
# Calculating cosine similarity
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
similarity.shape
```

```python
movies_list_filtered[movies_list_filtered['title'] == 'The Lego Movie'].index[0]
```

```python
#   Recommendation function
def recommend(movie):
    movie_index = movies_list_filtered[movies_list_filtered['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(movies_list_filtered.iloc[i[0]].title)
```

```python
recommend('Batman')
```

```python
# Saving the files using pickle
import pickle
movie_list = movies_list_filtered.reset_index(drop=True)

pickle.dump(movie_list, open('movies_dict.pkl','wb'))
pickle.dump(similarity, open('similarity.pkl','wb'))
```
